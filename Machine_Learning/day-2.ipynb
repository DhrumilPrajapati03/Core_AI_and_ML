## 1. Why Statistics Matters for ML (10 minutes)

**Ask the class first:** *"When we say a model 'learns from data,' what is it actually doing, mathematically?"*

**Definition/framing:**
> Every ML algorithm is, underneath, a statistical procedure — it's finding patterns, relationships, and distributions in data and using them to make predictions. If you don't understand the statistics, you can still call `.fit()` on a model, but you won't understand *why* it worked, *when* it will fail, or *how* to fix it when it does.

**Concrete examples of stats hiding inside ML you'll meet later this course:**

| Later Topic | Statistical Concept Underneath |
|---|---|
| Linear Regression (Day 6–8) | Mean, variance, minimizing squared error |
| Naive Bayes (Day 17) | Bayes' Theorem, conditional probability |
| Model evaluation (Day 10) | Distributions of correct/incorrect predictions |
| Feature scaling (Day 5, Day 22) | Standard deviation, normalization |
| Anomaly/outlier detection | Standard deviation, distribution tails |

**Simple analogy to use in class:**
> "If ML algorithms are recipes, statistics is understanding what 'a pinch of salt' or 'medium heat' actually means. You can follow a recipe blindly, but you can only *adjust* it — fix a dish that's too salty, scale it up for a bigger crowd — if you understand what's actually happening underneath."

## 2. Descriptive Statistics (25 minutes, hands-on)

### 2.1 Measures of Central Tendency

```python
import numpy as np
import pandas as pd

scores = np.array([56, 78, 92, 88, 45, 78, 90, 65, 78, 100])

mean_val = np.mean(scores)
median_val = np.median(scores)
mode_val = pd.Series(scores).mode()[0]

print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val)
```

| Measure | What it tells you | Weakness |
|---|---|---|
| **Mean** | The average value | Sensitive to outliers |
| **Median** | The "middle" value when sorted | Robust to outliers |
| **Mode** | The most frequent value | Not always meaningful for continuous data |

**Discussion prompt:** *"If your dataset is `[45, 50, 48, 52, 5000]`, which measure — mean or median — better represents a 'typical' value? Why?"* (Answer: median — the 5000 outlier drags the mean far from what's typical.)

### 2.2 Measures of Spread — Variance & Standard Deviation

```python
variance = np.var(scores)
std_dev = np.std(scores)

print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

**Intuition:**
> Mean tells you *where* the data is centered. Variance/std dev tell you *how spread out* it is. Two classes can have the same average score (mean=75) — but one class might have everyone scoring 70–80, and the other might range from 20–100. Same mean, very different reality.

**Simple analogy:** *"Mean is the bullseye you're aiming for. Standard deviation is how tightly your shots are clustered around it — a tight cluster (low std dev) means your predictions/measurements are consistent; a wide spread (high std dev) means they're all over the place."*

## 3. Distributions (20 minutes)

### 3.1 Why Distribution Shape Matters

> Knowing the mean and std dev isn't enough — the *shape* of your data matters too. Many ML algorithms (like Linear Regression and Naive Bayes) either assume data is roughly "Normal," or perform noticeably better/worse depending on the shape of the input.

### 3.2 The Normal (Gaussian) Distribution

```python
import matplotlib.pyplot as plt

data = np.random.normal(loc=70, scale=10, size=1000)  # mean=70, std=10
plt.hist(data, bins=30)
plt.title("Normal Distribution Example")
plt.show()
```

**Key properties to draw on the board (the classic bell curve):**
- Symmetric around the mean
- ~68% of data falls within 1 standard deviation of the mean
- ~95% falls within 2 standard deviations
- ~99.7% falls within 3 standard deviations

*(This is the "68-95-99.7 rule" — worth writing on the board exactly like that, since it comes back when discussing outliers and anomaly detection later.)*

### 3.3 Uniform & Skewed Distributions

```python
uniform_data = np.random.uniform(low=0, high=100, size=1000)
plt.hist(uniform_data, bins=30)
plt.title("Uniform Distribution Example")
plt.show()
```

| Distribution Shape | Example in Real Data |
|---|---|
| **Normal (bell curve)** | Heights, exam scores, measurement errors |
| **Uniform** | Random dice rolls, random ID assignment |
| **Right-skewed** (long tail right) | Income, house prices, most people's Instagram followers |
| **Left-skewed** (long tail left) | Age at retirement, exam scores capped near 100 |

**Discussion prompt:** *"If house prices are right-skewed (a few very expensive mansions pulling the tail), is the mean or median house price more representative of 'what a typical house costs'?"* (Answer: median — same reasoning as the outlier example earlier.)

## 4. Correlation (20 minutes)

### 4.1 What Correlation Measures

```python
import pandas as pd

df = pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "exam_score":  [50, 55, 60, 62, 70, 75, 85, 90]
})

correlation = df["study_hours"].corr(df["exam_score"])
print("Correlation:", correlation)
```

**Correlation coefficient range:** −1 to +1

| Value | Meaning |
|---|---|
| +1 | Perfect positive relationship (as X increases, Y increases perfectly) |
| 0 | No linear relationship |
| −1 | Perfect negative relationship (as X increases, Y decreases perfectly) |

### 4.2 Visualizing Correlation

```python
import seaborn as sns

sns.scatterplot(x="study_hours", y="exam_score", data=df)
plt.title("Study Hours vs Exam Score")
plt.show()
```

### 4.3 Correlation ≠ Causation (important — spend real time here)

**Classic example to use in class:** *"Ice cream sales and shark attacks are strongly correlated. Does ice cream cause shark attacks? Does one shark attack cause more ice cream sales?"*

> Both actually increase because of a hidden third factor: **hot summer weather**. More people swim (more shark exposure) AND more people buy ice cream. Neither causes the other — this is called a **confounding variable**.

**Why this matters for ML specifically:** A model can find that two variables are correlated and use that for prediction — but if you assume correlation implies you can *control* one variable to *change* the other, you can make seriously wrong real-world decisions (e.g., a hospital's ML model finding "patients who receive more treatment have worse outcomes" doesn't mean treatment causes worse outcomes — sicker patients receive more treatment in the first place).

### 4.4 Correlation Matrix (multiple variables at once)

```python
df2 = pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "sleep_hours": [8, 7, 7, 6, 6, 5, 5, 4],
    "exam_score":  [50, 55, 60, 62, 70, 75, 85, 90]
})

print(df2.corr())
sns.heatmap(df2.corr(), annot=True, cmap="coolwarm")
plt.show()
```

> This heatmap pattern will reappear constantly during EDA (Day 5) and feature selection (Day 22) — it's one of the first things any ML practitioner looks at on a new dataset.

## 5. Probability & Bayes' Theorem (25 minutes)

### 5.1 Basic Probability Refresher (quick, since this batch has some math background)

```
P(event) = number of favorable outcomes / total number of outcomes
```

Example: probability of rolling a 4 on a fair die = 1/6 ≈ 0.167

### 5.2 Conditional Probability

```
P(A | B) = probability of A happening, GIVEN that B has already happened
```

**Example:** P(it rains today | it's cloudy this morning) is very different from just P(it rains today) with no information at all. Conditioning on evidence changes the probability.

### 5.3 Bayes' Theorem

```
P(A | B) = [ P(B | A) × P(A) ] / P(B)
```

**Plain-language version to put on the board:**
> "The probability of A being true, given evidence B, equals: how likely you'd see evidence B if A were true, times how likely A was in the first place — all divided by how likely you'd see evidence B at all."

### 5.4 Worked Example — Medical Testing (the classic, very effective in class)

**Setup:**
- 1% of the population has a certain disease → `P(Disease) = 0.01`
- The test correctly identifies the disease 99% of the time → `P(Positive | Disease) = 0.99`
- The test has a 5% false positive rate → `P(Positive | No Disease) = 0.05`

**Question to pose to the class:** *"If someone tests positive, what's the actual probability they have the disease? Take a guess before we calculate."* (Most students will guess close to 95–99% — the reveal is the teaching moment.)

```python
p_disease = 0.01
p_no_disease = 0.99
p_positive_given_disease = 0.99
p_positive_given_no_disease = 0.05

# P(Positive) = P(Positive|Disease)*P(Disease) + P(Positive|No Disease)*P(No Disease)
p_positive = (p_positive_given_disease * p_disease) + (p_positive_given_no_disease * p_no_disease)

# Bayes' Theorem
p_disease_given_positive = (p_positive_given_disease * p_disease) / p_positive

print(f"P(Positive) = {p_positive:.4f}")
print(f"P(Disease | Positive) = {p_disease_given_positive:.4f}")
```

**Result:** ~16.7% — dramatically lower than most students guess.

**Why this happens (explain clearly):** Because the disease is so rare (1%), even a small false-positive rate (5%) generates far more false positives in absolute terms than true positives from the tiny infected population. This is one of the most important, most counter-intuitive results in applied statistics.

**Connect to ML directly:**
> "This exact reasoning is the Naive Bayes algorithm (Day 17) — and it's also why 'model accuracy' alone can be dangerously misleading on imbalanced datasets (Day 10, Day 23), where the rare class matters most but is easiest to miss."
