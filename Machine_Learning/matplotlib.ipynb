```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Define the range of input values (x)
x = np.linspace(-5, 5, 100)

# 2. Set a fixed weight (slope)
w = 2

# 3. Calculate y values for three different biases (b)
y_no_bias = w * x + 0  # b = 0
y_pos_bias = w * x + 3  # b = 3
y_neg_bias = w * x - 4  # b = -4

# 4. Create the plot
plt.figure(figsize=(10, 6))

# Plot the three lines with distinct colors and styles
plt.plot(
    x, y_no_bias, color="black", linestyle="--", label="No Bias (b = 0)"
)
plt.plot(
    x,
    y_pos_bias,
    color="green",
    linewidth=2,
    label="Positive Bias (b = +3) -> Shifts Up",
)
plt.plot(
    x,
    y_neg_bias,
    color="red",
    linewidth=2,
    label="Negative Bias (b = -4) -> Shifts Down",
)

# 5. Highlight the y-intercepts (where x = 0) to clearly show the shift
plt.scatter([0], [0], color="black", zorder=5)
plt.scatter([0], [3], color="green", zorder=5)
plt.scatter([0], [-4], color="red", zorder=5)

# Add text labels next to the intercept points
plt.text(0.2, 0, "(0, 0)", fontsize=10, color="black", weight="bold")
plt.text(0.2, 3, "(0, 3)", fontsize=10, color="green", weight="bold")
plt.text(0.2, -4, "(0, -4)", fontsize=10, color="red", weight="bold")

# 6. Style the chart for clear scannability
plt.axhline(0, color="gray", linewidth=0.8, linestyle=":")  # X-axis reference
plt.axvline(0, color="gray", linewidth=0.8, linestyle=":")  # Y-axis reference
plt.title("Visualizing the Bias Process ($y = wx + b$)", fontsize=14, pad=15)
plt.xlabel("Input ($x$)", fontsize=12)
plt.ylabel("Output ($y$)", fontsize=12)
plt.xlim(-5, 5)
plt.ylim(-10, 10)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11, loc="upper left")

# 7. Render the graph
plt.tight_layout()
plt.show()
```
