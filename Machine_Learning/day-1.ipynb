Machine Learning is a field of computer science where a system learns patterns from data and makes predictions or decisions without being explicitly programmed with rules for every scenario.

### Why Machine Learning Matters — Real-World Applications

### Types of Machine Learning

#### Supervised Learning

- Learns from labeled data (input + correct output given)
- Two sub-types:
    - Regression → predicts a continuous number (e.g., house price, temperature)
    - Classification → predicts a category (e.g., spam/not spam, disease/no disease)
- Examples: predicting house prices, email spam detection, loan approval prediction

#### Unsupervised Learning

- Learns from unlabeled data (no correct answer given) — finds hidden patterns/structure
- Sub-types:
    - Clustering → grouping similar data points (e.g., customer segmentation)
    - Dimensionality Reduction → simplifying data while keeping important information
- Examples: customer segmentation, anomaly detection, market basket analysis

#### Reinforcement Learning

- An agent learns by interacting with an environment, receiving rewards or penalties
- Examples: game-playing AI (Chess, Go, Atari), robotics, self-driving car decision-making

```python
from sklearn.datasets import load_iris
import pandas as pd

# Load a classic ML dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

print("Shape of dataset:", df.shape)
df.head()
```

```python
# A 3-line "teaser" model — don't explain the internals yet,
# just show that a machine can learn to classify flower species
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)

print("Model accuracy on unseen data:", model.score(X_test, y_test))
```
